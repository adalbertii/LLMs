```python
client = QdrantClient(os.getenv('QDRANT_HOST'), api_key=os.getenv('QDRANT_API_KEY'))
```

```python
# Load a sentence transformer model
model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
# Sample dataset
sentences = ["This is a sample sentence.", "Embeddings are useful.", "Sentence for the embedding"] # Add more sentences
```

```python
# Generate embeddings
start_time = time.time()
embeddings = model.encode(sentences)
end_time = time.time()
print("Time taken to generate embeddings:", end_time - start_time, "seconds")
```

```python
size_vector = len(embeddings[0])
print(size_vector)
```

```python
vector_param = VectorParams(size=len(embeddings[0]), distance=Distance.DOT)
client.create_collection(collection_name="wmi-01", vectors_config= vector_param)
client.upload_collection(collection_name="wmi-01", vectors=embeddings)
```

```python
quer_sentence = ["This is aa sampleee sentence."]
quer_emb = model.encode(quer_sentence)
```

```python
# Perform a search query
#query_vector = embeddings[0] # using the first sentence embedding as a query
search_results = client.search(collection_name="wmi-01", query_vector=quer_emb[0])
print(search_results)
```
