```python
client = QdrantClient(os.getenv('QDRANT_HOST'), api_key=os.getenv('QDRANT_API_KEY'))
```

```python
encoder = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
embedding_size = encoder.get_sentence_embedding_dimension()
print('Embedding vector size: ',embedding_size)
```

```python


# nazwa pliku do odczytu
plik = '/content/miejscowosci-01.csv'

# inicjujemy pustą listę documents
documents = []

# otwieramy plik do odczytu
with open(plik, 'r', encoding='utf-8') as f:
    for line in f:
        # usuwamy białe znaki na początku i końcu, w tym znaki nowej linii
        adres = line.strip()
        # tworzymy słownik i dodajemy do listy
        documents.append({"adres": adres})

# wyświetlamy wynik (opcjonalnie)

print(documents[10:15])
```

```python
collection = os.getenv('COLLECTION_NAME')

if not client.collection_exists(collection_name = collection):
  client.create_collection(
      collection_name=collection,
      vectors_config=models.VectorParams(
          size=embedding_size, # Vector size is defined by used model
          distance=models.Distance.COSINE
    )
  )


if client.collection_exists(collection_name = collection) :
  print('Collection: ', collection, ' has been created' )
```

```python
client.upload_points(
    collection_name=collection,
    points=[
        models.PointStruct(
            id=idx,
            vector=encoder.encode(doc["adres"]).tolist(),
            payload=doc
        ) for idx, doc in enumerate(documents)
    ]
)
```

```python
hits = client.query_points(
    collection_name=collection,
    query=encoder.encode("Adumka  69-480  al. Antaniego Roguckiego 17").tolist(),
    limit=3,
).points

for hit in hits:
    print(hit.payload, "score:", hit.score)
```
